```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
df.head()
```

```python
df_model = df.drop(["nameOrig", "nameDest", "isFlaggedFraud"], axis = 1)
```

```python
categorical = ["type"]
numeric = ["amount", "oldbalanceOrg", "newbalanceOrig", "oldbalanceDest", "newbalanceDest"]
```

```python
y = df_model["isFraud"] # target
X = df_model.drop("isFraud", axis = 1)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y)
```

```python
preprocessor = ColumnTransformer(                         # different preprocessing to different columns.
    transformers= [
        ("num", StandardScaler(), numeric),               # changes the scale so the values are centered around 0 with a standard deviation around 1
        ("cat", OneHotEncoder(drop="first"), categorical) # convert to numericals
    ],
    remainder= "drop"  # don't include unmentioned columns in the transformed data
)
```

```python
pipeline = Pipeline([
    ("prep", preprocessor),
    ("clf", LogisticRegression(class_weight="balanced", max_iter=1000)) # give equal omportance to Fraud as it is rare
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
predictions = pipeline.predict(X_test)
```

```python
print(classification_report(y_test, predictions))
```

```python
confusion_matrix(y_test, predictions)
```

```python
pipeline.score(X_test, y_test)
```

```python
# Export Pipeline

import joblib

joblib.dump(pipeline, "fraud_detection_pipeline.pkl")
```
